# Pandas Basics

##  Importing Pandas

Pandas is used for data analysis and working with tables.

```python
import pandas as pd

print("Pandas Imported Successfully")
```

##  Creating a DataFrame

A DataFrame is a table-like structure in Pandas.

```python
data = {
    "Name": ["Purva", "Aditya", "Rahul"],
    "Marks": [95, 88, 76]
}

df = pd.DataFrame(data)

print(df)
```

##  Accessing Columns

Pandas allows access to individual columns from a DataFrame.

```python
print(df["Name"])
```

##  Adding New Columns

New columns can be added easily to a DataFrame.

```python
df["Grade"] = ["A", "B", "C"]

print(df)
```

##  DataFrame Information

Pandas provides information about rows, columns, and data types.

```python
df.info()
```

##  Viewing First Rows

The head() function shows the first rows of a DataFrame.

```python
print(df.head())
```

##  Data Statistics

The describe() function provides statistical summary of numerical data.

```python
print(df.describe())
```

##  Reading CSV Files

Pandas can read CSV files and convert them into DataFrames.

```python
students_df = pd.read_csv("students.csv")

print(students_df)
```
