## 1. CREATE TABLE

CREATE TABLE is used to create a new table in a database.

Why it is useful:
Tables are used to store structured data such as student records, employee information, and product details.

```python
import sqlite3

conn = sqlite3.connect("students.db")

cursor = conn.cursor()

cursor.execute("""
CREATE TABLE students (
    id INTEGER,
    name TEXT,
    marks INTEGER
)
""")

conn.commit()

print("Table created successfully")
```

Explanation:

CREATE TABLE creates a new table called students.

Columns:
- id
- name
- marks

This table will store student information.

Key Concept:

CREATE TABLE table_name (
    column1 datatype,
    column2 datatype,
    column3 datatype

Summary:

✅ Created a database

✅ Created a table

✅ Defined columns and data types

## 2. INSERT INTO

INSERT INTO is used to add records (rows) into a table.

Why it is useful:
It allows us to store actual data inside the database.

```python
cursor.execute("""
INSERT INTO students VALUES
(1, 'Amit', 85)
""")

conn.commit()

print("Record inserted successfully")
```

```python
cursor.execute("""
INSERT INTO students VALUES
(2, 'Riya', 72)
""")

cursor.execute("""
INSERT INTO students VALUES
(3, 'John', 90)
""")

conn.commit()

print("More records inserted")
```

Explanation:

Three student records have been added to the students table.

Data Stored:

1, Amit, 85

2, Riya, 72

3, John, 90

Key Concept:

INSERT INTO table_name
VALUES (value1, value2, value3)

Summary:

✅ Inserted records into a table

✅ Stored student data

✅ Used INSERT INTO statement

## 3. SELECT

SELECT is used to retrieve data from a table.

Why it is useful:
It allows us to view and analyze stored information.

```python
cursor.execute("SELECT * FROM students")

rows = cursor.fetchall()

for row in rows:
    print(row)
```

Key Concept:

SELECT * FROM table_name

* means all columns.

Summary:

✅ Retrieved data from a table

✅ Displayed all records

✅ Used SELECT statement

## 4. WHERE Clause

The WHERE clause is used to filter records that match a specific condition.

Why it is useful:
It helps us retrieve only the required data instead of displaying the entire table.

### Example 1: Students with Marks Greater Than 80

```python
cursor.execute("SELECT * FROM students WHERE marks > 80")

rows = cursor.fetchall()

for row in rows:
    print(row)
```

Explanation:

Only students having marks greater than 80 are displayed.

85 → Included

90 → Included

72 → Not Included

### Example 2: Student with Name 'John'

```python
cursor.execute("SELECT * FROM students WHERE name = 'John'")

rows = cursor.fetchall()

for row in rows:
    print(row)
```

Key Concept:

SELECT * FROM table_name
WHERE condition

Examples:

WHERE marks > 80

WHERE marks < 70

WHERE name = 'John'

Summary:

✅ Filtered records using WHERE

✅ Retrieved specific rows

✅ Applied conditions in SQL queries
