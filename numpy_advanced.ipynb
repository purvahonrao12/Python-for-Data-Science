## 1. Reshape Arrays

Reshape is used to change the shape of an array without changing its data.

Why it is useful:
It helps organize data into rows and columns.

```python
import numpy as np

arr = np.array([1, 2, 3, 4, 5, 6])

arr.reshape(2, 3)
```

## 2. Mathematical Operations

NumPy allows mathematical operations on arrays.

Examples:
- Addition
- Subtraction
- Multiplication
- Division

```python
import numpy as np

a = np.array([10, 20, 30])

b = np.array([1, 2, 3])

print(a + b)
print(a - b)
print(a * b)
print(a / b)
```

## 3. Array Statistics

NumPy provides built-in functions for calculations.

```python
arr = np.array([10, 20, 30, 40, 50])

print("Sum:", np.sum(arr))
print("Mean:", np.mean(arr))
print("Maximum:", np.max(arr))
print("Minimum:", np.min(arr))
```
