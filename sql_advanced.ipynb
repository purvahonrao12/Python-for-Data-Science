```python
import sqlite3

conn = sqlite3.connect("school.db")

cursor = conn.cursor()

print("Database Connected")
```

```python
cursor.execute("INSERT INTO students VALUES (1, 'Amit', 85)")
cursor.execute("INSERT INTO students VALUES (2, 'John', 90)")
cursor.execute("INSERT INTO students VALUES (3, 'Sara', 78)")

conn.commit()

print("Data Inserted")
```

## Setup

We first create a database, create a students table, and insert sample data that will be used in the SQL Advanced examples below.

## 1. UPDATE

UPDATE is used to modify existing records in a table.

Why it is useful:
Sometimes data changes and we need to update the stored values.

```python
cursor.execute("""
UPDATE students
SET marks = 95
WHERE name = 'John'
""")

conn.commit()

print("Record updated successfully")
```

```python
cursor.execute("SELECT * FROM students")

rows = cursor.fetchall()

for row in rows:
    print(row)
```

Explanation:

John's marks were changed from 90 to 95.

Key Concept:

UPDATE table_name
SET column = value
WHERE condition

## 2. DELETE

DELETE is used to remove records from a table.

Why it is useful:
It helps remove unwanted or incorrect data.

```python
cursor.execute("""
DELETE FROM students
WHERE name = 'Riya'
""")

conn.commit()

print("Record deleted successfully")
```

```python
cursor.execute("SELECT * FROM students")

rows = cursor.fetchall()

for row in rows:
    print(row)
```

Explanation:

Riya's record has been removed from the table.

Key Concept:

DELETE FROM table_name
WHERE condition

## 3. GROUP BY

GROUP BY is used to group rows having the same value.

Why it is useful:
It helps summarize data and perform calculations on groups.

```python
cursor.execute("""
CREATE TABLE departments (
    department TEXT,
    marks INTEGER
)
""")

conn.commit()
```

```python
cursor.execute("INSERT INTO departments VALUES ('AI', 85)")
cursor.execute("INSERT INTO departments VALUES ('AI', 90)")
cursor.execute("INSERT INTO departments VALUES ('CS', 75)")
cursor.execute("INSERT INTO departments VALUES ('CS', 80)")

conn.commit()
```

```python
cursor.execute("""
SELECT department, AVG(marks)
FROM departments
GROUP BY department
""")

rows = cursor.fetchall()

for row in rows:
    print(row)
```

Explanation:

AI → Average Marks

CS → Average Marks

GROUP BY creates groups and AVG() calculates the average.

Key Concept:

SELECT column, AVG(column2)
FROM table
GROUP BY column

## 4. JOIN

JOIN is used to combine data from two or more tables based on a related column.

Why it is useful:
In real databases, information is often stored in multiple tables. JOIN helps us retrieve related data from those tables.

```python
cursor.execute("""
CREATE TABLE students_join (
    id INTEGER,
    name TEXT
)
""")

conn.commit()
```

```python
cursor.execute("INSERT INTO students_join VALUES (1, 'Amit')")
cursor.execute("INSERT INTO students_join VALUES (2, 'John')")

conn.commit()
```

```python
cursor.execute("""
CREATE TABLE marks (
    id INTEGER,
    marks INTEGER
)
""")

conn.commit()
```

```python
cursor.execute("INSERT INTO marks VALUES (1, 85)")
cursor.execute("INSERT INTO marks VALUES (2, 95)")

conn.commit()
```

### Example: INNER JOIN

We will combine the students_join table and marks table using the common column 'id'.

```python
cursor.execute("""
SELECT students_join.name, marks.marks
FROM students_join
INNER JOIN marks
ON students_join.id = marks.id
""")

rows = cursor.fetchall()

for row in rows:
    print(row)
```

Explanation:

The id column is common in both tables.

JOIN combines:
- Student name from students_join
- Marks from marks

Result:
Amit → 85
John → 95

Key Concept:

SELECT column_names
FROM table1
INNER JOIN table2
ON table1.common_column = table2.common_column
