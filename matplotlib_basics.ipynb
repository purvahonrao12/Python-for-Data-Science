```python
import matplotlib.pyplot as plt
import numpy as np
```

# 1. Line Plot

A line plot is used to visualize trends over time or continuous data.

```python
days = [1, 2, 3, 4, 5]
study_hours = [2, 4, 3, 5, 6]

plt.plot(days, study_hours)

plt.show()
```

```python
days = [1, 2, 3, 4, 5]
study_hours = [2, 4, 3, 5, 6]

plt.plot(days, study_hours)

plt.title("Study Hours Over 5 Days")
plt.xlabel("Days")
plt.ylabel("Study Hours")

plt.show()
```

# 2. Scatter Plot

Scatter plots show relationships between two variables.

```python
study_hours = [2, 3, 4, 5, 6]
marks = [60, 65, 70, 80, 90]

plt.scatter(study_hours, marks)

plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")

plt.show()
```

# 3. Bar Chart

Bar charts are used to compare categories.

```python
subjects = ["Python", "Maths", "Physics", "Chemistry"]
hours = [5, 3, 4, 2]

plt.bar(subjects, hours)

plt.title("Hours Spent Per Subject")
plt.xlabel("Subjects")
plt.ylabel("Hours")

plt.show()
```

# 4. Histogram

Histograms show the distribution of numerical data.

```python
marks = [55, 60, 62, 65, 68, 70, 72, 75, 80, 82, 85, 90]

plt.hist(marks)

plt.title("Distribution of Marks")
plt.xlabel("Marks")
plt.ylabel("Frequency")

plt.show()
```

# 5. Pie Chart

Pie charts show percentage contribution of categories.

```python
activities = ["Study", "Sleep", "Exercise", "Entertainment"]

hours = [8, 8, 2, 6]

plt.pie(hours, labels=activities, autopct="%1.1f%%")

plt.title("Daily Activities")

plt.show()
```

# 6. Legends

Legends help identify and distinguish multiple data series in a plot.

```python
days = [1, 2, 3, 4, 5]

python_hours = [1, 2, 2, 3, 4]
college_hours = [5, 5, 6, 5, 6]

plt.plot(days, python_hours, label="Python")
plt.plot(days, college_hours, label="College")

plt.legend()

plt.title("Python vs College Hours")

plt.show()
```

# 7. Grid

Grids improve plot readability by adding reference lines to the graph.

```python
days = [1, 2, 3, 4, 5]
study_hours = [2, 4, 3, 5, 6]

plt.plot(days, study_hours)

plt.grid(True)

plt.title("Study Hours Trend")

plt.show()
```

# Practice Exercise

```python
months = ["Jan", "Feb", "Mar", "Apr", "May"]

savings = [1000, 1500, 1200, 1800, 2200]

plt.plot(months, savings)

plt.title("Monthly Savings")
plt.xlabel("Month")
plt.ylabel("Amount")

plt.grid(True)

plt.show()
```
