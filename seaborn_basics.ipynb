# Seaborn Basics

## Topics Covered

- Countplot
- Boxplot
- Pairplot
- Heatmap
- Distribution Plot
- Correlation Visualization

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

# 1. Countplot

A countplot displays the frequency of each category in a dataset.

```python
data = pd.DataFrame({
    "Department": ["IT", "CS", "IT", "ENTC", "CS", "IT"]
})

sns.countplot(data=data, x="Department")

plt.title("Department Count")
plt.show()
```

# 2. Boxplot

A boxplot summarizes data distribution and helps identify outliers.

```python
data = pd.DataFrame({
    "Marks": [55, 60, 62, 65, 70, 75, 80, 95]
})

sns.boxplot(y=data["Marks"])

plt.title("Marks Distribution")
plt.show()
```

# 3. Pairplot

A pairplot visualizes relationships between multiple numerical variables.

```python
data = pd.DataFrame({
    "Study_Hours": [2,3,4,5,6],
    "Sleep_Hours": [8,7,7,6,6],
    "Marks": [60,65,70,80,90]
})

sns.pairplot(data)

plt.show()
```

# 4. Heatmap

A heatmap uses colors to represent data values in a matrix.

```python
data = pd.DataFrame({
    "Python": [80,85,90],
    "Maths": [75,82,88],
    "Physics": [70,78,84]
})

sns.heatmap(data, annot=True)

plt.title("Subject Scores Heatmap")
plt.show()
```

# 5. Distribution Plot

A distribution plot shows how data values are spread across a range.

```python
data = pd.DataFrame({
    "Marks": [55,60,62,65,70,75,80,85,90]
})

sns.histplot(data["Marks"], kde=True)

plt.title("Marks Distribution")
plt.show()
```

# 6. Correlation Visualization

Correlation visualization helps understand relationships between numerical variables.

```python
data = pd.DataFrame({
    "Study_Hours":[2,3,4,5,6],
    "Sleep_Hours":[8,7,7,6,6],
    "Marks":[60,65,70,80,90]
})

corr = data.corr()

sns.heatmap(corr, annot=True)

plt.title("Correlation Matrix")
plt.show()
```
