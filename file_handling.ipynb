# File Handling in Python

File Handling allows us to store and retrieve data from files.

Why it is useful:
- Store information permanently
- Read existing data
- Write new data
- Append data without deleting old content

Topics Covered:
- Reading Files
- Writing Files
- Appending Files

## 1. Writing to a File

Writing means creating a new file and storing data inside it.

Mode Used:
- "w" → Write Mode

Note:
If the file already exists, its old content will be replaced.

```python
with open("sample.txt", "w") as file:
    file.write("Hello, this is my first file.")

print("Data written successfully")
```

## 2. Reading a File

Reading allows us to view the contents stored inside a file.

Mode Used:
- "r" → Read Mode

```python
with open("sample.txt", "r") as file:
    content = file.read()

print(content)
```

## 3. Appending Data to a File

Appending means adding new data without deleting the existing content.

Mode Used:
- "a" → Append Mode

```python
with open("sample.txt", "a") as file:
    file.write("\nThis line was added later.")

print("Data appended successfully")
```

## 4. Reading Updated File

Let's verify that the new content was added successfully.

```python
with open("sample.txt", "r") as file:
    print(file.read())
```
