# NumPy Basics

##  Importing NumPy

NumPy is a Python library used for numerical and array operations.

```python
import numpy as np

print("NumPy Imported Successfully")
```

```python
numbers = np.array([1, 2, 3, 4, 5])

print(numbers)
```

##  Array Operations

NumPy allows fast mathematical operations on arrays.

```python
print(numbers + 2)
print(numbers * 2)
```

##  Array Indexing

Indexing is used to access specific elements from an array.

```python
print(numbers[0])
print(numbers[2])
```

##  Array Shape

The shape tells the dimensions of an array.

```python
print(numbers.shape)
```

##  Zeros Array

NumPy can create arrays filled with zeros.

```python
zeros_array = np.zeros(5)

print(zeros_array)
```

##  Ones Array

NumPy can create arrays filled with ones.

```python
ones_array = np.ones(5)

print(ones_array)
```

##  Random Arrays

NumPy can generate random numbers automatically.

```python
random_array = np.random.randint(1, 10, 5)

print(random_array)
```

##  Linspace

Linspace creates evenly spaced numbers between a range.

```python
line_array = np.linspace(0, 10, 5)

print(line_array)
```

##  2D Arrays

NumPy supports multi-dimensional arrays.

```python
matrix = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print(matrix)
```

```python
print(matrix[0][1])
```
